```python
# Generador de LUTs .cube para Super 16
# No requiere NumPy: usa solo la librería estándar de Python.

def generate_lut(filename, title, curve_func, size=33):
    with open(filename, 'w') as f:
        f.write(f'TITLE "{title}"\n')
        f.write(f'LUT_3D_SIZE {size}\n')
        f.write('DOMAIN_MIN 0.0 0.0 0.0\n')
        f.write('DOMAIN_MAX 1.0 1.0 1.0\n')
        for i in range(size):
            b = i / (size - 1)
            for j in range(size):
                g = j / (size - 1)
                for k in range(size):
                    r = k / (size - 1)
                    out = curve_func(r, g, b)
                    f.write(f'{out[0]:.6f} {out[1]:.6f} {out[2]:.6f}\n')

def unit01(r, g, b):
    r2 = min(max(r + 0.05 * (1 - r) + 0.02, 0), 1)
    g2 = min(max(g + 0.08 * g + 0.02, 0), 1)
    b2 = min(max(b + 0.10 * (1 - b) + 0.02, 0), 1)
    return (r2, g2, b2)

def unit02(r, g, b):
    r2 = min(max(r * 1.25 + 0.02, 0), 1)
    g2 = min(max(g * 1.15 + 0.01, 0), 1)
    b2 = min(max(b * 0.90, 0), 1)
    return (r2, g2, b2)

def base_neutral(r, g, b):
    r2 = min(max(0.5 + (r - 0.5) * 1.1, 0), 1)
    g2 = min(max(0.5 + (g - 0.5) * 1.05, 0), 1)
    b2 = min(max(0.5 + (b - 0.5) * 1.05, 0), 1)
    return (r2, g2, b2)

# Genera los 3 LUTs
generate_lut('EVA_Unit01.cube', 'EVA Unit01 PurpleGreen', unit01, size=33)
generate_lut('EVA_Unit02.cube', 'EVA Unit02 RedOrange', unit02, size=33)
generate_lut('Base_Film_Neutral.cube', 'Base Film Neutral', base_neutral, size=33)

print("✅ Archivos generados.")

# Descargar los archivos a tu ordenador
from google.colab import files
files.download('EVA_Unit01.cube')
files.download('EVA_Unit02.cube')
files.download('Base_Film_Neutral.cube')
```

```python
# Generador de LUTs Evangelion para Super 16mm Cam
# Ejecuta esta celda para generar y descargar los 9 LUTs restantes.

def generate_lut(filename, title, curve_func, size=33):
    with open(filename, 'w') as f:
        f.write(f'TITLE "{title}"\n')
        f.write(f'LUT_3D_SIZE {size}\n')
        f.write('DOMAIN_MIN 0.0 0.0 0.0\n')
        f.write('DOMAIN_MAX 1.0 1.0 1.0\n')
        for i in range(size):
            b = i / (size - 1)
            for j in range(size):
                g = j / (size - 1)
                for k in range(size):
                    r = k / (size - 1)
                    out = curve_func(r, g, b)
                    f.write(f'{out[0]:.6f} {out[1]:.6f} {out[2]:.6f}\n')

def clamp(val):
    return min(max(val, 0.0), 1.0)

# 1. Unit-00 (Azul/Naranja)
def unit00(r, g, b):
    # Azul frío en sombras, naranja cálido en luces
    r2 = clamp(r * 1.05 + 0.02)
    g2 = clamp(g * 1.00 + 0.01)
    b2 = clamp(b * 1.15 + 0.05 * (1 - b))
    return (r2, g2, b2)

# 2. NERV (Alerta/Emergencia)
def nerv(r, g, b):
    # Rojos agresivos, contraste alto
    r2 = clamp(0.5 + (r - 0.5) * 1.3 + 0.1)
    g2 = clamp(0.5 + (g - 0.5) * 1.2 + 0.02)
    b2 = clamp(0.5 + (b - 0.5) * 1.1 - 0.05)
    return (r2, g2, b2)

# 3. Tokyo-3 Crepúsculo
def tokyo3(r, g, b):
    # Ámbar en luces, púrpura en sombras
    r2 = clamp(r + 0.05 * (1 - r) + 0.02)
    g2 = clamp(g * 0.95)
    b2 = clamp(b + 0.10 * (1 - b))
    return (r2, g2, b2)

# 4. Interior Fluorescente
def fluorescente(r, g, b):
    # Verde enfermizo, bajo contraste
    r2 = clamp(0.5 + (r - 0.5) * 0.9)
    g2 = clamp(0.5 + (g - 0.5) * 0.95 + 0.05)
    b2 = clamp(0.5 + (b - 0.5) * 0.9)
    return (r2, g2, b2)

# 5. Nocturno EVA
def nocturno(r, g, b):
    # Sombras profundas, halation frío
    r2 = clamp(r * 0.95)
    g2 = clamp(g * 1.05)
    b2 = clamp(b * 1.1 + 0.05 * (1 - b))
    return (r2, g2, b2)

# 6. Alto Contraste Manga (B/N)
def manga_bw(r, g, b):
    # Convertir a blanco y negro puro con contraste extremo
    luma = 0.299*r + 0.587*g + 0.114*b
    contrast = clamp(0.5 + (luma - 0.5) * 1.5)
    return (contrast, contrast, contrast)

# 7. Manga Tramado (B/N)
def manga_dither(r, g, b):
    # B/N con negros levantados (estilo impresión de cómic)
    luma = 0.299*r + 0.587*g + 0.114*b
    contrast = clamp(0.05 + luma * 1.4)
    return (contrast, contrast, contrast)

# 8. Rojo/Naranja Suave
def red_orange(r, g, b):
    # Calidez cinematográfica
    r2 = clamp(r * 1.1 + 0.02)
    g2 = clamp(g * 1.05 + 0.01)
    b2 = clamp(b * 0.9)
    return (r2, g2, b2)

# 9. Azul/Blanco Frío
def blue_white(r, g, b):
    # Estética clínica y limpia
    r2 = clamp(r * 0.95)
    g2 = clamp(g * 0.98)
    b2 = clamp(b * 1.05 + 0.02)
    return (r2, g2, b2)

# Lista de LUTs a generar
luts_to_generate = [
    ("EVA_Unit00_BlueOrange.cube", "EVA Unit00 BlueOrange", unit00),
    ("EVA_NERV_Alert.cube", "EVA NERV Alert", nerv),
    ("EVA_Tokyo3_Twilight.cube", "EVA Tokyo3 Twilight", tokyo3),
    ("EVA_Fluorescent_Interior.cube", "EVA Fluorescent Interior", fluorescente),
    ("EVA_Night_Scene.cube", "EVA Night Scene", nocturno),
    ("EVA_Manga_HighContrast.cube", "EVA Manga HighContrast", manga_bw),
    ("EVA_Manga_Dither.cube", "EVA Manga Dither", manga_dither),
    ("EVA_RedOrange_Soft.cube", "EVA RedOrange Soft", red_orange),
    ("EVA_BlueWhite_Cold.cube", "EVA BlueWhite Cold", blue_white)
]

# Generar y descargar
print("Generando los 9 LUTs... Por favor espera.")
from google.colab import files

for filename, title, curve in luts_to_generate:
    generate_lut(filename, title, curve, size=33)
    files.download(filename)
    print(f"✅ Generado y descargado: {filename}")

print("\n🎉 ¡Todos los LUTs han sido generados y descargados con éxito!")
```

```python
# Generador de LUTs .cube para Super 16
# No requiere NumPy: usa solo la librería estándar de Python.

def generate_lut(filename, title, curve_func, size=33):
    with open(filename, 'w') as f:
        f.write(f'TITLE "{title}"\n')
        f.write(f'LUT_3D_SIZE {size}\n')
        f.write('DOMAIN_MIN 0.0 0.0 0.0\n')
        f.write('DOMAIN_MAX 1.0 1.0 1.0\n')
        for i in range(size):
            b = i / (size - 1)
            for j in range(size):
                g = j / (size - 1)
                for k in range(size):
                    r = k / (size - 1)
                    out = curve_func(r, g, b)
                    f.write(f'{out[0]:.6f} {out[1]:.6f} {out[2]:.6f}\n')

def unit01(r, g, b):
    r2 = min(max(r + 0.05 * (1 - r) + 0.02, 0), 1)
    g2 = min(max(g + 0.08 * g + 0.02, 0), 1)
    b2 = min(max(b + 0.10 * (1 - b) + 0.02, 0), 1)
    return (r2, g2, b2)

def unit02(r, g, b):
    r2 = min(max(r * 1.25 + 0.02, 0), 1)
    g2 = min(max(g * 1.15 + 0.01, 0), 1)
    b2 = min(max(b * 0.90, 0), 1)
    return (r2, g2, b2)

def base_neutral(r, g, b):
    r2 = min(max(0.5 + (r - 0.5) * 1.1, 0), 1)
    g2 = min(max(0.5 + (g - 0.5) * 1.05, 0), 1)
    b2 = min(max(0.5 + (b - 0.5) * 1.05, 0), 1)
    return (r2, g2, b2)

# Genera los 3 LUTs
generate_lut('EVA_Unit01.cube', 'EVA Unit01 PurpleGreen', unit01, size=33)
generate_lut('EVA_Unit02.cube', 'EVA Unit02 RedOrange', unit02, size=33)
generate_lut('Base_Film_Neutral.cube', 'Base Film Neutral', base_neutral, size=33)

print("✅ Archivos generados.")

# Descargar los archivos a tu ordenador
from google.colab import files
files.download('EVA_Unit01.cube')
files.download('EVA_Unit02.cube')
files.download('Base_Film_Neutral.cube')
```
